```python
import pandas as pd
import numpy as np

df = pd.read_excel(r"C:\Users\yashm\Desktop\AQI 2019-2023\Datasets\Air-Quality-dataset.xlsx")
df.head()
```

```python
df = df.rename(columns={'SO2 Annual Average': 'SO2', 
                        'NO2 Annual Average': 'NO2',
                        'PM10 Annual Average': 'PM10',
                        'PM2.5 Annual Average': 'PM2.5'})
df.head()
```

```python
df.columns = (
    df.columns.str.strip()
              .str.lower()
              .str.replace(" ", "_")
)
df.head()
```

```python
df["year"] = df["year"].astype("int")
pollutants = ["so2","no2","pm10","pm2.5"]
df[pollutants] = df[pollutants].apply(pd.to_numeric, errors = "coerce")
df.info()
df.isna().sum()
```

```python
df = df.dropna(subset = ["so2","no2","pm10"])
df.isna().sum()
```

```python
df["pm2.5"] = df.groupby("state/union_territory")["pm2.5"].transform(lambda x: x.fillna(x.median()))
df.isna().sum()
```

```python
df["state/union_territory"] = df["state/union_territory"].str.strip()
df["city/town/village"] = df["city/town/village"].str.strip()
```

```python
df.to_csv("clean_aqi_dataset.csv", index=False)
```
